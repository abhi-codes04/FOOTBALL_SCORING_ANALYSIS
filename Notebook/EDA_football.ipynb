# 1.ANALYZE THE TOP 10 GOAL SCORERS

```python
goal_scorers=df.groupby('Player Names')['Goals'].sum()
```

```python
top_goal_scorers = goal_scorers.sort_values(ascending=False)
```

```python
top_10 = top_goal_scorers.head(10)
```

```python
top_10
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(15,10))
```

```python
sns.barplot(x=top_10.values,y=top_10.index)
plt.title = 'top 10 scorers'
plt.xlabel = 'goals scored'
plt.ylabel = 'player names'
```

TASK 2 :- is to analyize goals scored per match by each player

or you can say ratio


```python
goals = df.groupby('Player Names')['Goals'].sum()
```

```python
matches = df.groupby('Player Names')['Matches_Played'].sum()
```

```python
Ratio = goals/matches
```

```python
top_10_byratio=Ratio.head(10)
```

```python
plt.figure(figsize=(20,15))
sns.barplot(x=top_10_byratio.index,y=top_10_byratio.values)
plt.title = 'TOP 10 BY GOALS RATIO'
```
